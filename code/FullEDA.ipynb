```python
from pathlib import Path
import pandas as pd
import json

folder = Path("../book_data/resumable")

summary = []
```

```python
for year in range(2006, 2026):
    csv_path = folder / f"checkouts_{year}.csv"
    progress_path = folder / f"progress_{year}.json"

    if csv_path.exists() and progress_path.exists():
        progress = json.loads(progress_path.read_text())

        rows = sum(
            len(chunk)
            for chunk in pd.read_csv(csv_path, chunksize=10000)
        )

        summary.append({
            "Year": year,
            "Rows": rows,
            "Download Complete": (
                progress["completed"] == len(progress["blocks"])
            )
        })

summary_df = pd.DataFrame(summary)
display(summary_df)

print("Total rows:", summary_df["Rows"].sum())
```

```python

import csv
from pathlib import Path

folder = Path("book_data/resumable")

results = []

for file in sorted(folder.glob("checkouts_*.csv")):
    with open(file, newline="", encoding="utf-8") as f:
        reader = csv.reader(f)
        header = next(reader)

        total = 0
        mismatched = 0

        for row in reader:
            total += 1
            if len(row) != len(header):
                mismatched += 1

    results.append({
        "Year": file.stem.replace("checkouts_", ""),
        "Rows": total,
        "Header Columns": len(header),
        "Mismatched Rows": mismatched
    })

import pandas as pd
display(pd.DataFrame(results))
```

```python
# Repair 2022 code


import csv
from pathlib import Path

source = Path("book_data/resumable/checkouts_2022.csv")
output_dir = Path("book_data/repaired")
output_dir.mkdir(parents=True, exist_ok=True)

destination = output_dir / "checkouts_2022.csv"

columns = [
    "usageclass", "checkouttype", "materialtype",
    "checkoutyear", "checkoutmonth", "checkouts",
    "title", "isbn", "creator", "subjects",
    "publisher", "publicationyear"
]

repaired = 0
preserved = 0

with open(source, newline="", encoding="utf-8") as infile, \
     open(destination, "w", newline="", encoding="utf-8") as outfile:

    reader = csv.reader(infile)
    writer = csv.writer(outfile)

    next(reader)
    writer.writerow(columns)

    for row in reader:
        if len(row) == 11:
            # Insert missing ISBN after title
            row.insert(7, "")
            repaired += 1

        elif len(row) == 12:
            preserved += 1

        else:
            raise ValueError(
                f"Unexpected column count: {len(row)}"
            )

        writer.writerow(row)

print(f"Rows with ISBN inserted: {repaired:,}")
print(f"Rows already 12 columns: {preserved:,}")
print(f"Total rows: {repaired + preserved:,}")
print(f"Saved to: {destination}")
```

```python
# Standardize the years


import csv
from pathlib import Path

source_dir = Path("book_data/resumable")
output_dir = Path("book_data/repaired")
output_dir.mkdir(exist_ok=True)

columns = [
    "usageclass", "checkouttype", "materialtype",
    "checkoutyear", "checkoutmonth", "checkouts",
    "title", "isbn", "creator", "subjects",
    "publisher", "publicationyear"
]

for year in list(range(2006, 2022)) + list(range(2023, 2026)):
    source = source_dir / f"checkouts_{year}.csv"
    destination = output_dir / f"checkouts_{year}.csv"

    with open(source, newline="", encoding="utf-8") as infile, \
         open(destination, "w", newline="", encoding="utf-8") as outfile:

        reader = csv.reader(infile)
        writer = csv.writer(outfile)

        next(reader)
        writer.writerow(columns)

        count = 0

        for row in reader:
            if len(row) == 11:
                row.insert(7, "")
            elif len(row) != 12:
                raise ValueError(
                    f"{year}: Unexpected {len(row)} columns"
                )

            writer.writerow(row)
            count += 1

    print(f"{year}: {count:,} rows repaired")
```

```python
# Verify alignment

import pandas as pd
from pathlib import Path

folder = Path("book_data/repaired")

expected_cols = [
    "usageclass", "checkouttype", "materialtype",
    "checkoutyear", "checkoutmonth", "checkouts",
    "title", "isbn", "creator", "subjects",
    "publisher", "publicationyear"
]

results = []

for year in range(2006, 2026):
    path = folder / f"checkouts_{year}.csv"

    df_year = pd.read_csv(
        path,
        dtype="string",
        low_memory=False
    )

    # Validate column names
    columns_ok = df_year.columns.tolist() == expected_cols

    # Convert numeric fields
    years = pd.to_numeric(
        df_year["checkoutyear"], errors="coerce"
    )
    months = pd.to_numeric(
        df_year["checkoutmonth"], errors="coerce"
    )
    checkouts = pd.to_numeric(
        df_year["checkouts"], errors="coerce"
    )

    # Check for invalid values
    bad_years = (~years.eq(year)).sum()
    bad_months = (~months.between(1, 12)).sum()
    bad_checkouts = (
        checkouts.isna() | (checkouts < 0) |
        (checkouts % 1 != 0)
    ).sum()

    # Publication year should generally be a year
    pub = df_year["publicationyear"].dropna()
    suspicious_pub = (
        ~pub.str.fullmatch(r"\d{4}")
    ).sum()

    # ISBN should be empty or look like an identifier
    isbn = df_year["isbn"].dropna()
    suspicious_isbn = (
        ~isbn.str.fullmatch(r"[\dXx\- ]{10,20}")
    ).sum()

    results.append({
        "Year": year,
        "Rows": len(df_year),
        "Columns OK": columns_ok,
        "Bad Years": bad_years,
        "Bad Months": bad_months,
        "Bad Checkouts": bad_checkouts,
        "Suspicious ISBN": suspicious_isbn,
        "Suspicious Publication Year": suspicious_pub
    })

results_df = pd.DataFrame(results)
display(results_df)

print("\nTOTAL ROWS:", results_df["Rows"].sum())
```

```python

import pandas as pd
from pathlib import Path

folder = Path("book_data/repaired")

for year in [2015, 2022, 2023, 2025]:
    df = pd.read_csv(
        folder / f"checkouts_{year}.csv",
        dtype="string"
    )

    print(f"\n{'=' * 20} YEAR {year} {'=' * 20}")

    # Show suspicious ISBN values
    isbn = df["isbn"].dropna()
    suspicious_isbn = isbn[
        ~isbn.str.fullmatch(r"[\dXx\- ]{10,20}")
    ]

    print("\nSuspicious ISBN examples:")
    print(suspicious_isbn.head(8).tolist())

    # Show suspicious publication years
    pub = df["publicationyear"].dropna()
    suspicious_pub = pub[
        ~pub.str.fullmatch(r"\d{4}")
    ]

    print("\nSuspicious publication year examples:")
    print(suspicious_pub.head(8).tolist())

    # Show full metadata rows where ISBN looks suspicious
    if not suspicious_isbn.empty:
        print("\nFull rows with suspicious ISBN:")
        display(
            df.loc[
                suspicious_isbn.index[:5],
                ["title", "isbn", "creator",
                 "subjects", "publisher", "publicationyear"]
            ]
        )
```
